```python
def find_innermost(input_list, depth=0):
    deepest_list = input_list
    max_depth = depth

    for item in input_list:
        if isinstance(item, list):
            candidate, candidate_depth = find_innermost(item, depth + 1)

            if candidate_depth > max_depth:
                deepest_list = candidate
                max_depth = candidate_depth

    return deepest_list, max_depth

def innermost_plus_one(input_list):
    deepest_list, _ = find_innermost(input_list)
    return [number + 1 for number in deepest_list]

input_list = [1, 2, 3, 4, [5, 6, 7, [8, 9]]]

result = innermost_plus_one(input_list)
print(result)
```
