```python
# 3.2 Choose y = 20: indices 0, 20, ..., 380 are within the dataset.
def twenty_spaced_values(dataset, y):
    if not isinstance(y, int) or y <= 0:
        raise ValueError("y must be a positive integer.")
    if 19 * y >= len(dataset):
        raise ValueError("The dataset is too short for 20 values at this spacing.")
    result = []
    for index in range(0, 20 * y, y):
        result.append(dataset[index])
    return result

y = 20
answer_3_2 = twenty_spaced_values(influenza_genome, y)
for value in answer_3_2:
    print(value)
```
