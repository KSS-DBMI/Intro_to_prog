```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Importing Data.
flights_data = pd.read_csv("flights.csv")
flights_data.head(10)
weather_data_pd = pd.read_csv("weather.csv")
weather_data_np = weather_data_pd.to_numpy()
```

#Question 1 How many flights were there from JFK to SLC? Int
q_1 

```python
# Each matching row represents one flight record.
jfk_slc = flights_data[(flights_data["origin"] == "JFK") &
                       (flights_data["dest"] == "SLC")]
q_1 = int(jfk_slc.shape[0])
q_1
```

#Question 2 How many airlines fly to SLC? Should be int
q_2 

```python
# Count airlines, not flight records.
q_2 = int(flights_data.loc[flights_data["dest"] == "SLC", "carrier"].nunique())
q_2
```

#Question 3 What is the average arrival delay for flights to RDU? float
q_3

```python
# Calculate the mean arrival delay for the requested destination.
q_3 = float(flights_data.loc[flights_data["dest"] == "RDU", "arr_delay"].mean())
q_3
```

#Question 4 What proportion of flights to SEA come from the two NYC airports (LGA and JFK)?  float
q_4 

```python
sea_flights = flights_data[flights_data["dest"] == "SEA"]
# True values count as 1, so their mean is the requested proportion.
q_4 = float(sea_flights["origin"].isin(["LGA", "JFK"]).mean())
q_4
```

#Question 5 Which date has the largest average depature delay? Pd slice with date and float
#please make date a column. Preferred format is 2013/1/1 (y/m/d)
q_5

```python
# Build a date column without leading zeroes, as requested.
flights_data["date"] = (flights_data["year"].astype(str) + "/" +
                        flights_data["month"].astype(str) + "/" +
                        flights_data["day"].astype(str))
# Average the available departure delays for each date.
daily_dep = flights_data.groupby("date", as_index=False)["dep_delay"].mean()
q_5 = daily_dep.loc[daily_dep["dep_delay"] == daily_dep["dep_delay"].max(),
                    ["date", "dep_delay"]]
q_5
```

#Question 6 Which date has the largest average arrival delay? pd slice with date and float
q_6

```python
# Group by date, then select the maximum average arrival delay.
daily_arr = flights_data.groupby("date", as_index=False)["arr_delay"].mean()
q_6 = daily_arr.loc[daily_arr["arr_delay"] == daily_arr["arr_delay"].max(),
                    ["date", "arr_delay"]]
q_6
```

#Question 7 Which flight departing LGA or JFK in 2013 flew the fastest? pd slice with tailnumber and speed
#speed = distance/airtime
q_7

```python
# Keep eligible flights with positive, nonmissing air time.
eligible_flights = flights_data.loc[
    (flights_data["origin"].isin(["LGA", "JFK"])) &
    (flights_data["year"] == 2013) &
    (flights_data["air_time"] > 0)
].copy()
eligible_flights["speed"] = (eligible_flights["distance"] /
                            eligible_flights["air_time"])
# Return the requested tail number and speed, including ties if any.
q_7 = eligible_flights.loc[
    eligible_flights["speed"] == eligible_flights["speed"].max(),
    ["tailnum", "speed"]
]
q_7
```

#Question 9 How many observations were made in Feburary? Int
q_9

```python
# Locate the month and humidity columns in the supplied array.
month_col = weather_data_pd.columns.get_loc("month")
humid_col = weather_data_pd.columns.get_loc("humid")
february_mask = weather_data_np[:, month_col] == 2
q_9 = int(np.count_nonzero(february_mask))
q_9
```

#Question 10 What was the mean for humidity in February? Float
q_10

```python
# Extract February humidity from the original NumPy array.
february_humidity = weather_data_np[february_mask, humid_col].astype(float)
q_10 = float(np.nanmean(february_humidity))
q_10
```

#Question 11 What was the std for humidity in February? Float
q_11

```python
# Exclude missing values and explicitly use NumPy's default ddof.
q_11 = float(np.nanstd(february_humidity, ddof=0))
q_11
```
