```python
# 2.a: Shared processing function with the corrected numeric section. A separate name keeps the diagnostic function from 1.a available.
def add_function(arg1, arg2):
    # numeric section
    if sum([type(i) == int for i in arg1]) == len(arg1) and \
       sum([type(i) == int for i in arg2]) == len(arg2):
        if len(arg1) != len(arg2):
            raise ValueError("Numeric input lists must have the same length.")
        arg1_index = 0
        while arg1_index < len(arg1):
            arg_2_sum = arg1[arg1_index] + arg2[arg1_index]
            arg1[arg1_index] = arg_2_sum
            arg1_index += 1
        return arg1
    #string section (original logic unchanged)
    elif sum([type(i)==str for i in arg1])==len(arg1) and \
        sum([type(i)==str for i in arg2])==len(arg2):
            arg1_index=0
            while arg1_index < len(arg1):
                arg_2_sum = ''
                for arg2_elements in arg2:
                    arg_2_sum += arg2_elements
                arg1[arg1_index]=arg1[arg1_index]+str(arg_2_sum)
                arg1_index+=1
            return arg1

answer_2a = add_function([1, 2, 3], [1, 1, 1])
print(answer_2a)
```

```python
# 2.b: Validate inputs and return an informative error message.
def exception_add_function(arg1, arg2):
    try:
        expected_type = int
        for value in arg1 + arg2:
            if type(value) == str:
                expected_type = str

        for argument_number, values in enumerate([arg1, arg2], start=1):
            for element_index, value in enumerate(values):
                if type(value) != expected_type:
                    raise TypeError(
                        f"Your input argument {argument_number} at element "
                        f"{element_index} is not of the expected type. "
                        "Please change this and rerun."
                    )
        return add_function(arg1.copy(), arg2.copy())
    except TypeError as error:
        return str(error)
    except ValueError as error:
        return str(error)

arg_str_1 = ['1', '2', '3']
arg_str_2 = ['1', '1', 1]
answer_2b = exception_add_function(arg_str_1, arg_str_2)
answer_2b_numeric = exception_add_function([1, 2, 3], [1, 1, 1])
answer_2b_other_example = exception_add_function(["5", "2", 5], ["1", "1", "1"])
print(answer_2b)
print(answer_2b_numeric)
print(answer_2b_other_example)
```

```python
# 2.c: Recover from mixed types inside the function.
def correction_add_function(arg1, arg2):
    try:
        all_integers = all(type(value) == int for value in arg1 + arg2)
        all_strings = all(type(value) == str for value in arg1 + arg2)
        if not all_integers and not all_strings:
            raise TypeError("Input types are mixed or unsupported.")
        return add_function(arg1.copy(), arg2.copy())
    except TypeError:
        # Convert local copies; the user's original lists stay unchanged.
        arg1 = [str(value) for value in arg1]
        arg2 = [str(value) for value in arg2]
        return add_function(arg1, arg2)
    except ValueError as error:
        return str(error)

answer_2c = correction_add_function(arg_str_1, arg_str_2)
answer_2c_numeric = correction_add_function([1, 2, 3], [1, 1, 1])
answer_2c_strings = correction_add_function(['1', '2', '3'], ['1', '1', '1'])
print(answer_2c)
print(answer_2c_numeric)
print(answer_2c_strings)
```
