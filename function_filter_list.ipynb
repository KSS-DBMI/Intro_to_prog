```python
def filter_numbers(numbers, threshold):
    filtered_list = []

    for number in numbers:
        if number <= threshold:
            filtered_list.append(number)

    return filtered_list

# Data
input_list = [3, 8, 1, 10, 5, 2]
threshold = 5

result = filter_numbers(input_list, threshold)
print(result)
```
